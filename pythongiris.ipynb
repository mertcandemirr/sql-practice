```python
class ogrenci :
    def __init__(self,isim,yas,notlar):
        self.isim=isim
        self.yas=yas
        self.notlar=notlar
    
    def ortalama_hesapla(self):
        return sum(self.notlar)/len(self.notlar)

    def durum(self):
        ort= self.ortalama_hesapla()
        return "Geçti" if ort>= 60 else "Kaldı"
```

```python
class Urun:
    def __init__(self, isim, fiyat, stok_adedi):
        self.isim = isim
        self.fiyat = fiyat
        self.stok_adedi = stok_adedi

    def toplam_hesapla(self):
        return self.fiyat * self.stok_adedi

    def stok_durumu(self):
        if self.stok_adedi == 0:
            return "Tükendi"
        elif self.stok_adedi < 10:
            return "Stok Azalmış"
        else:
            return "Yeterli Stok"

        
```
